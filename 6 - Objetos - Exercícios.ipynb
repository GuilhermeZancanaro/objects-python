1. Crie uma classe chamada Carro que tenha os seguintes atributos:
   * marca (string)
   * modelo (string)
   * ano (inteiro)
* Adicione um método chamado exibir_informacoes, que exiba as informações do carro.
* Crie dois objetos da classe Carro, preencha os valores e exiba as informações de cada objeto.

```python
# resolva o exercício aqui

class Carro:
    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano

    def __str__(self):
        return f"Marca: {self.marca}\nModelo: {self.modelo}\nAno: {self.ano}\n"

    def exibir_informaçoes(self):
        print(self.__str__())

carro1 = Carro("BYD","Dolphin mini", 2026)
carro1.exibir_informaçoes()

carro2 = Carro("Chevrolet", "Spin", 2020)
carro2.exibir_informaçoes()
        
```

2. Crie uma classe chamada Retangulo com os atributos:
   * largura (float)
   * altura (float)

* Adicione os seguintes métodos:
  * calcular_area – que retorna a área do retângulo.
  * calcular_perimetro – que retorna o perímetro do retângulo.

* Crie um objeto da classe Retangulo, defina os valores de largura e altura e exiba a área e o perímetro chamando os métodos.

```python
# resolva o exercício aqui
class Retangulo:
    def __init__(self, largura, altura):
        self.largura = float(largura)
        self.altura = float(altura)

    def caclcular_area(self):
        return self.largura * self.altura

    def calcular_perimetro(self):
        return 2 * (self.largura + self.altura)


    def __str__(self):
        return f"A area do retangulo é: {self.caclcular_area()}, E o perimtro do retangulo é: {self.calcular_perimetro()}"

    def resultado(self):
        return self.__str__()

calculo_retangulo = Retangulo(5.3, 2.5)
calculo_retangulo.resultado()



```

3. Crie uma classe chamada ContaBancaria com os seguintes atributos privados:
   * titular (string)
   * saldo (float)

* Adicione os seguintes métodos:
	1.	depositar(valor) – que aumenta o saldo da conta.
	2.	sacar(valor) – que diminui o saldo da conta, apenas se houver saldo suficiente.
	3.	exibir_saldo() – que exibe o saldo atual.

* Crie um objeto da classe ContaBancaria, faça depósitos, saques e exiba o saldo após cada operação.

```python
# resolva o exercício aqui
class ContaBancaria:

    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo

    def depositar(self,valor):
        self.saldo += valor

    def sacar(self,valor):
        if valor <= self.saldo:
            self.saldo -= valor
        else:
            print("saldo insuficiente")

    def saldo_atual(self):
        return print(f"O seu saldo é de R$: {self.saldo}")

conta = ContaBancaria("André", 5000.0)

conta.saldo_atual()
conta.depositar(200.0)
conta.saldo_atual()
conta.sacar(100.0)
conta.saldo_atual()
conta.sacar(15000.0)
conta.saldo_atual()
```

4. Crie uma classe chamada Aluno com os atributos:
   * nome (string)
   * nota (float)

* Crie uma classe chamada Turma, que possui:
  * Um atributo chamado alunos, que será uma lista de objetos da classe Aluno.
  * Um método adicionar_aluno(aluno) – que adiciona um objeto do tipo Aluno à lista.
  * Um método calcular_media – que retorna a média das notas de todos os alunos da turma.
  * Um método exibir_alunos - que exibe todos os alunos e suas respectivas notas da turma.

1. Crie três objetos da classe Aluno com nomes e notas à sua escolha.
2. Adicione esses objetos a um objeto da classe Turma.
3. Calcule e exiba a média das notas da turma.
4. Exiba os alunos da turma e suas notas.

```python
# resolva o exercício aqui
class Aluno:
    def __init__(self, nome, nota):
        self.nome = nome
        self.nota = nota

class Turma:

    def __init__(self):
        self.alunos = []

    def adicionar_aluno(self, aluno):
        self.alunos.append(aluno)

    def exibir_alunos(self):
        for aluno in self.aluno:   
            print(f"Aluno: {aluno.nome}, Nota: {aluno.nota}")

    def calcular_media(self):
        total_notas = sum(aluno.nota for aluno in self.alunos)
        return total_notas / len((self.alunos)) if self.alunos else 0

aluno1 = Aluno("antonia", 5.2)
aluno2 = Aluno("jose", 8.8)
aluno3 = Aluno("mario", 9.7)
    
turma = Turma()


turma.adicionar_aluno(aluno1)
turma.adicionar_aluno(aluno2)
turma.adicionar_aluno(aluno3)


media = turma.calcular_media()

print(f"media de notas da turma: {round(media, 2)}")
```
